```python

import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL_NAME = "google/flan-t5-small"

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)
print("Loading model...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)
model = model.to(device)
model.eval()

print("StudyMate AI model loaded successfully!")
```

```python

def generate_text(prompt, max_new_tokens=180):
    """Generate text from an instruction and return the result."""
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(device)

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            num_beams=2,
            do_sample=False,
            early_stopping=True
        )

    return tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )


def prepare_notes(notes):
    """Validate and clean the student's input."""
    if not notes or not notes.strip():
        raise ValueError("Please enter your lecture notes first.")

    return notes.strip()


def summarize_notes(notes):
    notes = prepare_notes(notes)
    prompt = (
        "Summarize the following lecture notes into clear, concise "
        "study notes. Include the main ideas and important definitions. "
        "Do not invent facts.\n\n"
        f"Lecture notes:\n{notes}"
    )
    return generate_text(prompt, max_new_tokens=180)


def make_flashcards(notes):
    notes = prepare_notes(notes)
    prompt = (
        "Create 5 study flashcards from the following lecture notes. "
        "For each flashcard, write a question and its answer. "
        "Use only information from the notes. Format as numbered "
        "Question and Answer pairs.\n\n"
        f"Lecture notes:\n{notes}"
    )
    return generate_text(prompt, max_new_tokens=220)


def make_quiz(notes):
    notes = prepare_notes(notes)
    prompt = (
        "Create 3 multiple-choice quiz questions based on these notes. "
        "Each question must have options A, B, C and D, followed by "
        "the correct answer and a brief explanation. "
        "Use only information from the notes.\n\n"
        f"Lecture notes:\n{notes}"
    )
    return generate_text(prompt, max_new_tokens=250)


print("All three study functions are ready!")
```
